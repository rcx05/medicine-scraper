```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import json
import time

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}


LABELS = [

    "Active Ingredient",
    "Active Ingredients",

    "Salt Composition",
    "Composition",

    "Key Ingredients",
    "Key Ingredient",

    "Ingredients",

    "Generic Name",
    "Generic",

    "Salt"
]

STOP_WORDS = [

    "Uses",
    "Benefits",
    "Side effects",
    "Side Effects",
    "Safety Advice",
    "Storage",
    "Manufacturer",
    "Product introduction",
    "How to use",
    "Directions",
    "Quick Tips",
    "Interactions",
    "Drug Warnings",
    "FAQs",
    "Disclaimer"

]



def universal_extract(soup):

    # ------------------------------------------------------
    # JSON-LD
    # ------------------------------------------------------

    scripts = soup.find_all("script", type="application/ld+json")

    for script in scripts:

        try:

            data = json.loads(script.string)

            if isinstance(data, dict):

                for key in [

                    "activeIngredient",
                    "activeIngredients",
                    "ingredient",
                    "ingredients"

                ]:

                    if key in data:

                        return str(data[key]).strip()

        except:
            pass


    text = soup.get_text("\n", strip=True)



    for label in LABELS:

        pattern = (

            re.escape(label)

            +

            r"\s*(.*?)\s*(?="

            +

            "|".join(STOP_WORDS)

            +

            r"|$)"

        )

        m = re.search(

            pattern,

            text,

            flags=re.I | re.S

        )

        if m:

            ans = m.group(1)

            ans = ans.replace("\n", " ")

            ans = re.sub(r"\s+", " ", ans)

            ans = ans.strip()

            if len(ans) > 2:

                return ans



    patterns = [

        r'Each tablet contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)',

        r'Each capsule contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)',

        r'Each softgel contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)',

        r'Each sachet contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)',

        r'Each serving contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)',

    ]

    text2 = soup.get_text(" ", strip=True)

    for p in patterns:

        m = re.search(

            p,

            text2,

            flags=re.I | re.S

        )

        if m:

            return m.group(1).strip()



    meta = soup.find("meta", attrs={"name":"description"})

    if meta:

        desc = meta.get("content","")

        m = re.search(

            r'contains (.*?)(?:\.|$)',

            desc,

            flags=re.I

        )

        if m:

            return m.group(1).strip()

    return None
```

```python


def scrape_1mg_generic(soup):

    ans = universal_extract(soup)

    if ans:
        return ans

    text = soup.get_text("\n", strip=True)

    idx = text.find("Also known as")

    if idx != -1:

        lines = text[idx:].split("\n")

        if len(lines) >= 2:
            return lines[1].strip()

    h1 = soup.find("h1")

    if h1:
        return h1.text.strip()

    return None




def scrape_1mg_drug(soup):

    ans = universal_extract(soup)

    if ans:
        return ans

    return None




def scrape_apollo(soup):

    ans = universal_extract(soup)

    if ans:
        return ans

    return None



def scrape_netmeds(soup):

    ans = universal_extract(soup)

    if ans:
        return ans

    return None




def get_active_ingredient(url):

    if pd.isna(url):
        return ""

    url = str(url).strip()

    if url == "":
        return ""

    if not url.startswith("http"):
        return ""

    try:

        r = requests.get(
            url,
            headers=HEADERS,
            timeout=30
        )

        if r.status_code != 200:
            return ""

        soup = BeautifulSoup(
            r.text,
            "lxml"
        )

        # --------------------------
        # 1MG GENERIC
        # --------------------------

        if "1mg.com/generics/" in url:

            return scrape_1mg_generic(soup)

        # --------------------------
        # 1MG DRUG
        # --------------------------

        elif "1mg.com/drugs/" in url:

            return scrape_1mg_drug(soup)

        # --------------------------
        # APOLLO
        # --------------------------

        elif "apollopharmacy.in" in url:

            return scrape_apollo(soup)

        # --------------------------
        # NETMEDS
        # --------------------------

        elif "netmeds.com" in url:

            return scrape_netmeds(soup)

        else:

            return ""

    except Exception as e:

        print(e)

        return ""
```

```python
# ==========================================================
# READ EXCEL
# ==========================================================

FILE_NAME = "manual_to_mapped_soap1.xlsx"

URL_COLUMN = "url"

df = pd.read_excel(FILE_NAME)

print("Columns Found:")
print(df.columns.tolist())

# Remove spaces from column names
df.columns = df.columns.str.strip()

ingredients = []

TOTAL = len(df)

for i, url in enumerate(df[URL_COLUMN], start=1):

    print(f"{i}/{TOTAL}")

    ingredient = None

    # Retry twice
    for attempt in range(2):

        ingredient = get_active_ingredient(url)

        if ingredient not in [None, "", "Not Found"]:
            break

        time.sleep(1)

    if ingredient is None or ingredient == "":
        ingredient = "Not Found"

    ingredients.append(ingredient)

    time.sleep(0.5)


df["Active Ingredient"] = ingredients

OUTPUT = "output_active_ingredients1.xlsx"

df.to_excel(
    OUTPUT,
    index=False
)

print("\nDONE")
print("Saved as:", OUTPUT)
```
