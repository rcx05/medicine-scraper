```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}


# -------------------------------------------------------
# Extract from 1mg Generic page
# -------------------------------------------------------
def scrape_1mg_generic(soup):

    text = soup.get_text("\n", strip=True)

    idx = text.find("Also known as")

    if idx != -1:
        lines = text[idx:].split("\n")

        if len(lines) >= 2:
            return lines[1].strip()

    h1 = soup.find("h1")
    if h1:
        return h1.text.strip()

    return None


# -------------------------------------------------------
# Extract from 1mg Drug page
# -------------------------------------------------------
def scrape_1mg_drug(soup):

    text = soup.get_text("\n", strip=True)

    match = re.search(
        r"Salt Composition\s*(.*?)\s*(Product introduction|Uses|Benefits|Side effects)",
        text,
        re.S,
    )

    if match:
        ingredient = match.group(1)
        ingredient = ingredient.replace("\n", " ").strip()

        return ingredient

    return None


# -------------------------------------------------------
# Apollo
# -------------------------------------------------------
def scrape_apollo(soup):

    text = soup.get_text("\n", strip=True)

    m = re.search(
        r"Composition\s*(.*?)\s*(Manufacturer|Uses|Benefits|Storage)",
        text,
        re.S,
    )

    if m:
        return m.group(1).replace("\n", " ").strip()

    return None


# -------------------------------------------------------
# Netmeds
# -------------------------------------------------------
def scrape_netmeds(soup):

    text = soup.get_text("\n", strip=True)

    m = re.search(
        r"Generic Name\s*(.*?)\s*(Manufacturer|Introduction|Uses)",
        text,
        re.S,
    )

    if m:
        return m.group(1).replace("\n", " ").strip()

    return None


# -------------------------------------------------------
# Master Function
# -------------------------------------------------------
def get_active_ingredient(url):

    if pd.isna(url):
        return ""

    url = str(url).strip()

    if url == "":
        return ""

    if not url.startswith("http"):
        return ""

    try:

        r = requests.get(url, headers=HEADERS, timeout=20)

        if r.status_code != 200:
            return ""

        soup = BeautifulSoup(r.text, "lxml")

        if "1mg.com/generics/" in url:
            return scrape_1mg_generic(soup)

        elif "1mg.com/drugs/" in url:
            return scrape_1mg_drug(soup)

        elif "apollopharmacy.in" in url:
            return scrape_apollo(soup)

        elif "netmeds.com" in url:
            return scrape_netmeds(soup)

        else:
            return ""

    except Exception:
        return ""


# ============================================================
# READ EXCEL
# ============================================================

df = pd.read_excel("manual_to_mapped_soap.xlsx")

print(df.columns)

# CHANGE THIS COLUMN NAME
URL_COLUMN = "url"

ingredients = []

for i, url in enumerate(df[URL_COLUMN], start=1):

    print(f"{i}/{len(df)}")

    ingredient = get_active_ingredient(url)

    ingredients.append(ingredient)

    time.sleep(0.5)

df["Active Ingredient"] = ingredients

df.to_excel("output_active_ingredients.xlsx", index=False)

print("DONE")
```
