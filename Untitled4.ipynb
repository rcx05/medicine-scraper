```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import json
import time

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By


# ============================================================
# FILE
# ============================================================

INPUT_FILE = "manual_to_mapped_soap1.xlsx"
OUTPUT_FILE = "manual_to_mapped_soap1_with_active_ingredients.xlsx"

URL_COLUMN = "url"

df = pd.read_excel(INPUT_FILE)

df.columns = df.columns.str.strip()

print("Total rows:", len(df))
print("URLs:", df[URL_COLUMN].notna().sum())


# ============================================================
# LABELS
# ============================================================

LABELS = [
    "Active Ingredient",
    "Active Ingredients",
    "Salt Composition",
    "Composition",
    "Key Ingredients",
    "Key Ingredient",
    "Ingredients",
    "Generic Name",
    "Generic",
    "Salt"
]

STOP_WORDS = [
    "Uses",
    "Benefits",
    "Side Effects",
    "Side effects",
    "Safety Advice",
    "Storage",
    "Manufacturer",
    "Product introduction",
    "How to use",
    "Directions",
    "Drug Warnings",
    "Interactions",
    "FAQs",
    "Disclaimer"
]


# ============================================================
# CLEAN TEXT
# ============================================================

def clean(text):

    text = re.sub(r"\s+", " ", str(text))

    return text.strip()


# ============================================================
# EXTRACT FROM PAGE TEXT
# ============================================================

def extract_from_text(text):

    lines = [
        clean(x)
        for x in text.split("\n")
        if clean(x)
    ]

    # --------------------------------------------------------
    # LABEL BASED
    # --------------------------------------------------------

    for i, line in enumerate(lines):

        for label in LABELS:

            if line.lower() == label.lower():

                collected = []

                for j in range(i + 1, min(i + 10, len(lines))):

                    nxt = lines[j]

                    if nxt.lower() in [x.lower() for x in LABELS]:
                        break

                    if any(
                        stop.lower() in nxt.lower()
                        for stop in STOP_WORDS
                    ):
                        break

                    collected.append(nxt)

                if collected:

                    result = clean(" ".join(collected))

                    if result:

                        return result


    # --------------------------------------------------------
    # "Each tablet contains"
    # --------------------------------------------------------

    patterns = [

        r"Each tablet contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)",

        r"Each capsule contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)",

        r"Each softgel contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)",

        r"Each serving contains[:\s]*(.*?)(?:Uses|Benefits|Side Effects|Safety Advice|Disclaimer|$)"

    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            flags=re.I | re.S
        )

        if match:

            result = clean(match.group(1))

            if result:
                return result


    return None


# ============================================================
# SELENIUM
# ============================================================

options = Options()

options.add_argument("--headless=new")
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")

driver = webdriver.Chrome(options=options)


# ============================================================
# SCRAPER
# ============================================================

def scrape_url(url):

    if pd.isna(url):
        return "Not Found"

    url = str(url).strip()

    if not url.startswith("http"):
        return "Invalid URL"

    try:

        driver.get(url)

        time.sleep(2)

        # Get fully rendered page
        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        result = extract_from_text(page_text)

        if result:
            return result

        # ----------------------------------------------------
        # HTML fallback
        # ----------------------------------------------------

        html = driver.page_source

        soup = BeautifulSoup(
            html,
            "html.parser"
        )

        # JSON-LD
        for script in soup.find_all(
            "script",
            type="application/ld+json"
        ):

            try:

                data = json.loads(script.string)

                if isinstance(data, dict):

                    for key in [
                        "activeIngredient",
                        "activeIngredients",
                        "ingredient",
                        "ingredients"
                    ]:

                        if key in data:

                            return clean(data[key])

            except:
                pass


        result = extract_from_text(
            soup.get_text("\n", strip=True)
        )

        if result:
            return result


        return "Not Found"


    except Exception as e:

        print("ERROR:", url)
        print(e)

        return "Error"


# ============================================================
# PROCESS ALL URLS
# ============================================================

results = []

total = len(df)

for i, url in enumerate(df[URL_COLUMN], start=1):

    print(f"{i}/{total}  |  {url}")

    result = scrape_url(url)

    results.append(result)

    # Small delay
    time.sleep(0.5)


# ============================================================
# SAVE
# ============================================================

df["Active Ingredient"] = results

df.to_excel(
    OUTPUT_FILE,
    index=False
)

driver.quit()

print("\n====================================")
print("DONE")
print("====================================")
print("Output:", OUTPUT_FILE)

print("\nResult distribution:")
print(df["Active Ingredient"].value_counts().head(20))
```
