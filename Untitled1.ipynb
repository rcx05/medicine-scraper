```python
#import the medicine data that we stored earlier

import pickle

with open("all_medicines.pkl", "rb") as f:
    medicines = pickle.load(f)

print(len(medicines))
```

```python
#scrape the medicine name and id based on the same salt

from collections import defaultdict

salt_groups = defaultdict(list)

for medicine in medicines:

    salt = medicine["salt"]

    salt_groups[salt].append({
        "id": medicine["id"],
        "name": medicine["name"]
    })
```

```python
#save the data to pickle form

import pickle

with open("medicines_by_salt.pkl", "wb") as f:
    pickle.dump(salt_groups, f)

print("Done!")
```

```python
#save the data to json form

import json
with open("medicines_by_salt.json", "w", encoding="utf-8") as f:
    json.dump(dict(salt_groups), f, indent=4, ensure_ascii=False)

print("Done!")
```

```python
#print total unique salts and also number of medicines consisting of a given salt

print(f"Total unique salts: {len(salt_groups)}")
print()

sorted_salts = sorted(
    salt_groups.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

for salt, medicine_list in sorted_salts:
    print(f"{salt} : {len(medicine_list)} medicines")
```

```python
# separate salt name and strength and give a id to each salt so that later we can distinguish that the salt was for a particular medicine

import pandas as pd
import re

rows = []

for medicine in medicines:

    brand = medicine["name"]
    salt_string = medicine["salt"]

    components = salt_string.split(" + ")

    for component in components:

        match = re.match(r"(.+?)\s*\((.*?)\)", component)

        if match:

            rows.append({
                "Brand": brand,
                "Salt": match.group(1).strip(),
                "Strength": match.group(2).strip()
            })

df_brand_salt = pd.DataFrame(rows)

print(df_brand_salt.shape)

print(df_brand_salt.head(20))
```

```python
#Analyzes Paracetamol (325mg) combinations and calculates how many brands exist for medicines containing 1 salt, 2 salts, 3 salts, and larger salt combinations.

from collections import Counter

salt_count_distribution = Counter()

target = "Paracetamol (325mg)"

for salt_string, medicine_list in salt_groups.items():

    if target in salt_string:

        num_salts = len(salt_string.split(" + "))

        salt_count_distribution[num_salts] += len(medicine_list)

for num_salts, brand_count in sorted(salt_count_distribution.items()):

    print(
        f"{num_salts} salt(s) : {brand_count} brands"
    )
```

```python
#conversion to dataframe

import pandas as pd

salt_distribution_df = pd.DataFrame(
    salt_count_distribution.items(),
    columns=["Number of Salts", "Brand Count"]
)

print(salt_distribution_df)
```

```python
#creates a dot plot for the brand count per each salt

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    salt_distribution_df["Number of Salts"],
    salt_distribution_df["Brand Count"],
    s=100
)

plt.xlabel("Number of Salts in Combination")
plt.ylabel("Number of Brands")
plt.title("Paracetamol (325mg) Brand Distribution by Combination Size")

plt.grid(alpha=0.3)

plt.show()
```

```python
#Counts the number of salts present in each medicine, stores the counts in a DataFrame, and displays the frequency of medicines having 1, 2, 3, or more salts.

import pandas as pd
import matplotlib.pyplot as plt

salt_counts = []

for medicine in medicines:

    salt_string = medicine["salt"]

    if salt_string.strip():

        num_salts = len(salt_string.split(" + "))
        salt_counts.append(num_salts)

df_salt_per_medicine = pd.DataFrame({
    "Number of Salts": salt_counts
})

print(df_salt_per_medicine["Number of Salts"].value_counts().sort_index())
```

```python
#Builds a network graph where each node represents a salt, and an edge connects two salts if they appear together in the same medicine formulation.

import networkx as nx

G = nx.Graph()

for salt_string in salt_groups.keys():

    components = salt_string.split(" + ")

    salts = []

    for component in components:

        if "(" in component:

            salt_name = component.split("(")[0].strip()

            salts.append(salt_name)

    for i in range(len(salts)):
        for j in range(i + 1, len(salts)):

            G.add_edge(
                salts[i],
                salts[j]
            )
```

```python
#Calculates the number of connections for each salt in the network, sorts them in descending order, and displays the 20 most connected salts.

degree_df = sorted(
    G.degree(),
    key=lambda x: x[1],
    reverse=True
)

for salt, degree in degree_df[:20]:
    print(f"{salt}: {degree} connections")
```

```python
#stores the complete record in csv form

import pandas as pd

degree_df_csv = pd.DataFrame(
    degree_df,
    columns=["Salt", "Connections"]
)

degree_df_csv.to_csv(
    "salt_network_connections.csv",
    index=False
)

print("Saved!")
```

```python
#save dataframe as pickle

degree_df_csv.to_pickle(
    "degree_df.pkl"
)

print("Done!")
```
